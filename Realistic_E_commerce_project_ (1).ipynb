### Interpretation of the Summary Statistics

This table provides a comprehensive summary of your DataFrame, `df`.

**For Numerical Columns (e.g., Age, Unit Price, Quantity, Total Price, Shipping Fee):**

*   **count**: Number of non-null observations. This helps identify columns with missing values.
*   **mean**: The average value.
*   **std**: The standard deviation, which measures the amount of variation or dispersion of a set of values.
*   **min**: The smallest value.
*   **25% (Q1)**: The 25th percentile (first quartile).
*   **50% (Median/Q2)**: The 50th percentile (median or second quartile).
*   **75% (Q3)**: The 75th percentile (third quartile).
*   **max**: The largest value.

**For Categorical Columns (e.g., Customer ID, Gender, Region, Product Name, Category, Shipping Status, Order Date):**

*   **count**: Number of non-null observations.
*   **unique**: The number of distinct values in the column.
*   **top**: The most frequently occurring value.
*   **freq**: The frequency of the `top` value.

```python
sns.countplot(
    x='Region',
    hue='Gender',
    data=df,
    palette='viridis',
    order=['South','West','North','East']
)
plt.title('Region vs Gender')
plt.show()
```

The count plot illustrates the distribution of individuals across four regions (South, West, North, and East) categorized by gender. The visualization helps compare male and female representation in each region. From the plot, we can observe differences in gender distribution across regions and identify which region has the highest participation.

```python
from matplotlib.transforms import Bbox
from pandas.core.base import NoNewAttributesMixin
# bar-plot
bar=sns.barplot(
    x='Category',
    y='Total Price',
    hue='Gender',
    data=df,
    estimator='mean',
    palette='Set2',
    errorbar=None
)

plt.title('Average Total Price by Category and Gender')
plt.legend(bbox_to_anchor=(1,1))
plt.bar_label(bar.containers[0])
plt.bar_label(bar.containers[1])
plt.show()
```

## Interpretation of your above chart

The bar plot displays the average total price for different product categories segmented by gender. It allows comparison of purchasing behavior between male and female customers across categories. From the visualization, we can identify which category has the highest average spending and observe gender-based differences in purchasing patterns.

```python
sns.scatterplot(
    data=df,
    x='Total Price',
    y='Quantity',
    hue='Gender',
    palette='bright'
)
plt.legend(bbox_to_anchor=(1,1))
plt.title('Total Price vs Quantity')
plt.grid(True)
plt.show()
```

## Interpretation of your above chart

The scatter plot visualizes the relationship between total price and quantity purchased. The plot indicates a positive relationship, where higher quantities generally lead to higher total prices. Additionally, the color grouping by gender helps compare purchasing patterns between male and female customers.

```python
sns.scatterplot(
    data=df,
    x='Total Price',
    y='Quantity',
    hue='Gender',
    palette='bright',
    size='Category',
    style='Region'

)
plt.legend(bbox_to_anchor=(1,1))
plt.title('Total Price vs Quantity')
plt.grid(True)
plt.show()
```

## Interpretation of your above char

The scatter plot illustrates the relationship between total price and quantity purchased. The visualization shows a positive relationship where higher quantities generally correspond to higher total prices. Color differentiation by gender allows comparison of purchasing behavior between male and female customers, while marker size and style highlight variations across product categories and regions

```python
sns.histplot(data=df, x='Total Price', hue='Category', kde=True)
plt.title('Distribution of Total Price by Category')
plt.legend(bbox_to_anchor=(1,1))
plt.xlabel('Total Price')
plt.ylabel('Count')
plt.show()
```

## Interpretation of your above chart

The histogram visualizes the distribution of total purchase prices across different product categories. The plot shows how frequently transactions occur within specific price ranges. By separating the distribution by category, we can compare pricing patterns across product groups. The KDE curve highlights the overall trend and helps identify the most common purchase price ranges.

```python
sns.violinplot(
    x="Category",
    y="Unit Price",
    data=df,
    split=True,          # split by hue -- True
    inner="quartile",    # box, quartile, point, stick
    palette="Set2"
)
plt.title("Violin Plot of Products by Price and Category")
plt.legend(bbox_to_anchor=(1,1))
plt.show()
```

The violin plot illustrates the distribution of unit prices across different product categories. The width of each violin represents the density of product prices within that category, while the quartile lines highlight the central price distribution. This visualization helps identify categories with higher price ranges and greater price variability.
